```python
#B22DCCN476 - Đỗ Ngọc Lâm
#Cell [1]. [Import NumPy]

import numpy as np

print("NumPy version:", np.__version__)
```

```python
#B22DCCN476 - Đỗ Ngọc Lâm
#Cell [2]. [Scalar – Tensor 0 chiều]

# Scalar (0D tensor)
a = np.array(5)

print("a =", a)
print("Số chiều (ndim):", a.ndim)
print("Shape:", a.shape)
```

```python
#B22DCCN476 - Đỗ Ngọc Lâm
#Cell [3]. [Vector – Tensor 1 chiều]

# Vector (1D tensor)
v = np.array([1, 2, 3, 4])

print("v =", v)
print("Số chiều (ndim):", v.ndim)
print("Shape:", v.shape)
```

```python
#B22DCCN476 - Đỗ Ngọc Lâm
#Cell [4]. [Matrix – Tensor 2 chiều]

# Matrix (2D tensor)
m = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("m =\n", m)
print("Số chiều (ndim):", m.ndim)
print("Shape:", m.shape)
```

```python
#B22DCCN476 - Đỗ Ngọc Lâm
#Cell [5]. [Tensor 3 chiều – ví dụ ảnh xám]

# Tensor 3D: (số ảnh, chiều cao, chiều rộng)
images = np.random.randint(
    0, 256,
    size=(2, 4, 4),
    dtype=np.uint8
)

print("images =\n", images)
print("Số chiều (ndim):", images.ndim)
print("Shape:", images.shape)
```

```python
#B22DCCN476 - Đỗ Ngọc Lâm
#Cell [6]. [Tensor 4 chiều – ảnh màu (batch, H, W, C)]

# Tensor 4D: batch ảnh RGB
rgb_images = np.random.randint(
    0, 256,
    size=(2, 4, 4, 3),
    dtype=np.uint8
)

print("Shape:", rgb_images.shape)
print("Số chiều (ndim):", rgb_images.ndim)
```

```python
#B22DCCN476 - Đỗ Ngọc Lâm
#Cell [7]. [Phép toán trên tensor (element-wise)]

x = np.array([[1, 2],
              [3, 4]])

y = np.array([[10, 20],
              [30, 40]])

print("x + y =\n", x + y)
print("x * y =\n", x * y)
```

```python
#B22DCCN476 - Đỗ Ngọc Lâm
#Cell [8]. [Broadcasting – tính chất quan trọng của tensor]

a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.array([10, 20, 30])

print("a shape:", a.shape)
print("b shape:", b.shape)
print("a + b =\n", a + b)
```
